```python
import sqlalchemy
import pymysql
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

df = pd.read_csv("../data/cleaned_food_delivery_data.csv")

connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password="apple",
    host="localhost",
    port=3306,
    database="online_food_delivery"

    )

engine = create_engine(connection_url)

with engine.connect() as connection:
    print("Connected to online_food_delivery database!")
```

```python
with engine.connect() as connection:
    connection.exec_driver_sql(
        "CREATE DATABASE IF NOT EXISTS online_food_delivery"
    )

print("Database created successfully!")
```

```python
df.dtypes
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

df["Order_Date"].dtype
```

```python
from sqlalchemy.types import VARCHAR, Integer, DECIMAL, Date

df.to_sql(
    "orders",
    con=engine,
    if_exists="replace",
    index=False,
    dtype={
        "Order_ID": VARCHAR(50),
        "Customer_ID": VARCHAR(50),
        "Customer_Age": Integer(),
        "Customer_Gender": VARCHAR(20),
        "City": VARCHAR(50),
        "Area": VARCHAR(100),
        "Restaurant_ID": VARCHAR(50),
        "Restaurant_Name": VARCHAR(100),
        "Cuisine_Type": VARCHAR(50),
        "Order_Date": Date(),
        "Delivery_Time_Min": Integer(),
        "Distance_km": DECIMAL(10, 2),
        "Order_Value": DECIMAL(10, 2),
        "Discount_Applied": DECIMAL(10, 2),
        "Final_Amount": DECIMAL(10, 2),
        "Payment_Mode": VARCHAR(30),
        "Order_Status": VARCHAR(30),
        "Cancellation_Reason": VARCHAR(50),
        "Delivery_Partner_ID": VARCHAR(50),
        "Delivery_Rating": DECIMAL(3, 2),
        "Restaurant_Rating": DECIMAL(3, 2),
        "Order_Day": VARCHAR(20),
        "Day_of_Week": VARCHAR(20),
        "Peak_Hour": VARCHAR(20),
        "Profit_Margin": DECIMAL(5, 2),
        "Profit_Margin_Percentage": DECIMAL(6, 2),
        "Delivery_Performance": VARCHAR(20),
        "Customer_Age_Group": VARCHAR(30)
    }
)

print("Data successfully stored in MySQL!")
```

```python
with engine.connect() as connection:
    result = connection.exec_driver_sql("DESCRIBE orders")
    
    for row in result:
        print(row)
```

```python
with engine.connect() as connection:
    result = connection.exec_driver_sql("SELECT COUNT(*) FROM orders")
    print("Total rows:", result.fetchone()[0])
```
