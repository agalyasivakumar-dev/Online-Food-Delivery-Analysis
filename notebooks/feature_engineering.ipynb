```python
import pandas as pd
```

```python
df = pd.read_csv("../data/cleaned_food_delivery_data.csv")
df.head()
```

```python
# Profit margin percentage

df["Profit_Margin"].head(10)
```

```python
df["Profit_Margin_Percentage"] = df["Profit_Margin"] * 100
```

```python
df[["Profit_Margin", "Profit_Margin_Percentage"]].head(10)
```

```python
# Delivery performance categories

def delivery_category(time):
    if time <= 58:
        return "Fast"
    elif time <= 165:
        return "Moderate"
    else:
        return "Slow"

df["Delivery_Performance"] = df["Delivery_Time_Min"].apply(delivery_category)
```

```python
df["Delivery_Performance"].value_counts()
```

```python
# Customer age groups
def age_group(age):
    if age <= 25:
        return "Young Adult"
    elif age <= 35:
        return "Adult"
    elif age <= 50:
        return "Middle Age"
    else:
        return "Senior"
df["Customer_Age_Group"] = df["Customer_Age"].apply(age_group)
```

```python
df["Customer_Age_Group"].value_counts()
```

```python
#creating the actual day of the week from Order_Date
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Day_of_Week"] = df["Order_Date"].dt.day_name()
```

```python
df["Day_of_Week"].value_counts()
```
