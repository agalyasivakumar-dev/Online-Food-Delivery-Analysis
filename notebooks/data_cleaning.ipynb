```python
import pandas as pd
```

```python
df = pd.read_csv("../data/ONINE_FOOD_DELIVERY_ANALYSIS.csv")
df.head()
```

```python
df.duplicated().sum()
```

```python
# Checking duplicates in Order ID
df["Order_ID"].duplicated().sum()
```

```python
# Dropping Order time since it has no values 
df = df.drop(columns=["Order_Time"])
```

```python
# Handling the below category values with unKnown
categorical_columns = [
    "Customer_Gender",
    "City",
    "Area",
    "Cuisine_Type",
    "Payment_Mode"
]

for col in categorical_columns:
    df[col] = df[col].fillna("Unknown")
```

```python
# Using median to fill all the missing values for the customer Age 
df["Customer_Age"] = df["Customer_Age"].fillna(df["Customer_Age"].median())
```

```python
#Making sure that there are any outliers in the Delivery Time
Q1 = df["Delivery_Time_Min"].quantile(0.25)
Q3 = df["Delivery_Time_Min"].quantile(0.75)
IQR = Q3 - Q1
LB = Q1- 1.5 * IQR
UB = Q3 + 1.5 * IQR
Outliers = df[(df["Delivery_Time_Min"] < LB) | (df["Delivery_Time_Min"] > UB)]
print("Number of outliers:", len(Outliers))
```

```python
df["Delivery_Time_Min"] = pd.to_numeric(
    df["Delivery_Time_Min"], errors="coerce"
)

df["Delivery_Time_Min"] = df["Delivery_Time_Min"].fillna(
    df["Delivery_Time_Min"].median()
)
```

```python
#Making sure that there are any outliers in the Distance KM
Q1 = df["Distance_km"].quantile(0.25)
Q3 = df["Distance_km"].quantile(0.75)
IQR = Q3 - Q1
LB = Q1- 1.5 * IQR
UB = Q3 + 1.5 * IQR
Outliers = df[(df["Distance_km"] < LB) | (df["Distance_km"] > UB)]
print("Number of outliers:", len(Outliers))
```

```python
df["Distance_km"] = df["Distance_km"].fillna(df["Distance_km"].median())
```

```python
#Checking Outliers for order value
Q1 = df["Order_Value"].quantile(0.25)
Q3 = df["Order_Value"].quantile(0.75)
IQR = Q3 - Q1
LB = Q1- 1.5 * IQR
UB = Q3 + 1.5 * IQR
Outliers = df[(df["Order_Value"] < LB) | (df["Order_Value"] > UB)]
print("Number of outliers:", len(Outliers))
```

```python
df["Order_Value"] = df["Order_Value"].fillna(df["Order_Value"].median())
```

```python
#Cancellation_Reason
df["Cancellation_Reason"].value_counts(dropna=False)
```

```python
#not every cancelled order has a cancellation reason. Some of those may be delivered orders. so comparing order status with cancellation reason
pd.crosstab(df["Order_Status"], df["Cancellation_Reason"], dropna=False)
```

```python
#For cancelled orders where the reason is missing, we'll use Unknown.
#For delivered orders where the reason is missing, we'll use Not Cancelled.
df.loc[
    (df["Order_Status"] == "Cancelled") & (df["Cancellation_Reason"].isna()),
    "Cancellation_Reason"
] = "Unknown"

df.loc[
    (df["Order_Status"] == "Delivered") & (df["Cancellation_Reason"].isna()),
    "Cancellation_Reason"
] = "Not Cancelled"
```

```python
df["Cancellation_Reason"].value_counts(dropna=False)
```

```python
# Restaurant_Rating
df["Restaurant_Rating"].describe()
```

```python
(df["Restaurant_Rating"] > 5).sum()
```

```python
# Usually ratings are expected to be in a range of 1 - 5, here we have values which are more than 5. So capping those values with 5 
df["Restaurant_Rating"] = df["Restaurant_Rating"].clip(upper=5)
```

```python
df["Restaurant_Rating"].max()
```

```python
# Peak Hour
df["Peak_Hour"].value_counts(dropna=False)
```

```python
df["Peak_Hour"].dtypes
```

```python
# Since Peak_Hour is a Boolean field, we shouldn't fill the missing values with "Unknown" because that would change its data type.
df["Peak_Hour"] = df["Peak_Hour"].map({
    True: "Peak Hour",
    False: "Non-Peak Hour"
}).fillna("Unknown")
```

```python
# Delivery Rating 
df["Delivery_Rating"].value_counts(dropna=False)
```

```python
# Rating is not applicable when an order is cancelled
df.groupby("Order_Status")["Delivery_Rating"].agg(
    ["count", "mean", "min", "max"]
)
```

```python
#For all cancelled orders, set the Delivery_Rating value to None
df.loc[df["Order_Status"] == "Cancelled", "Delivery_Rating"] = None

# Finding Median median of delivered ratings only.
delivered_median = df.loc[
    df["Order_Status"] == "Delivered", "Delivery_Rating"
].median()

delivered_median
```

```python
# Filling missing Delivery_Rating values only for delivered orders
df.loc[
    (df["Order_Status"] == "Delivered") &
    (df["Delivery_Rating"].isna()),
    "Delivery_Rating"
] = delivered_median
```

```python
## Cancelled → NaN because delivery rating doesn't apply.
## Delivered → actual ratings retained
## Missing delivered ratings → filled with the median of delivered ratings only.
df.groupby("Order_Status")["Delivery_Rating"].agg(
    ["count", "mean", "min", "max"]
)
```

```python
df["Discount_Applied"].describe()
```

```python
df["Discount_Applied"].isna().sum()
```

```python
#Making sure that there are any outliers in the Discounts
Q1 = df["Discount_Applied"].quantile(0.25)
Q3 = df["Discount_Applied"].quantile(0.75)
IQR = Q3 - Q1
LB = Q1- 1.5 * IQR
UB = Q3 + 1.5 * IQR
Outliers = df[(df["Discount_Applied"] < LB) | (df["Discount_Applied"] > UB)]
print("Number of outliers:", len(Outliers))
```

```python
# Missing discount values were filled with the median discount of 50 because some discount values were unusually high.
df["Discount_Applied"] = df["Discount_Applied"].fillna(df["Discount_Applied"].median())
```
